```python
import requests
import json
import os
from tqdm import tqdm
import time
import random
```

```python
mean = 2.0
std_dev = 0.5
least = 0.5
def random_wait():
    """ 按高斯分布随机等待时间
    """
    wait_time = max(least, random.gauss(mean, std_dev))
    time.sleep(wait_time)
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.5',
    'Referer': 'https://www.google.com/',  # 模拟从 Google 跳转
    'Connection': 'keep-alive',
}
```

```python
with open('../data/song_info.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
save_prefix_path = '../data/img_song_cover/'
```

```python
step = 0
failed = 0
for index, [id, song] in enumerate(tqdm(data.items(), desc='下载歌曲封面图片')):
    if index + 1 < step:
        continue
    
    img_url = song['cover_img_url']
    try:
        response = requests.get(img_url, stream=True)
        response.raise_for_status()

        with open(save_prefix_path + id + '.jpg', 'wb') as f:
            # 分块下载
            for chunk in response.iter_content(1024):
                f.write(chunk)
    except Exception:
        failed += 1
    
    random_wait()

print(f'有 {failed} 个图片下载失败')
        
```

```python
with open('../data/artist_info.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

save_prefix_path = '../data/img_artist/'
```

```python
step = 0
failed = 0
queue = []
for index, [id, artist] in enumerate(tqdm(data.items(), desc='下载艺术家图片')):
    if index + 1 < step:
        continue
    img_url = artist['img_url']
    save_path = save_prefix_path + id + '.jpg'
    if os.path.exists(save_path):
        continue
    try:
        response = requests.get(img_url, stream=True)
        response.raise_for_status()

        with open(save_path, 'wb') as f:
            # 分块下载
            for chunk in response.iter_content(1024):
                f.write(chunk)
    except Exception:
        failed += 1
        queue.append([img_url, save_path])
    
    random_wait()

print(f'有 {failed} 个图片下载失败')
```
